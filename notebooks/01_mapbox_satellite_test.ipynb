# Mapbox Satellite Image Test

This notebook downloads one satellite image from Mapbox for Vienna center and displays it.

The image will be saved to `data/mapbox_satellite_vienna.png`.

This notebook requests a `512x512@2x` Mapbox image, which produces a `1024x1024` pixel file for the later pixel-to-lon/lat conversion step.

```python
# Import the libraries we need.
# sys lets us add the project folder to Python's import path.
# Path helps us build file paths that work on different operating systems.
import sys
from pathlib import Path
```

```python
# Find the project root folder.
# If this notebook is opened from the notebooks/ folder, the project root is one level up.
# If it is opened from the project root, the current folder is already the project root.
current_dir = Path.cwd()

if (current_dir / ".env").exists():
    project_root = current_dir
else:
    project_root = current_dir.parent

# Add the project root to Python's import path.
# This lets the notebook import code from backend/app/imagery.
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

# The downloader function will read MAPBOX_TOKEN from the .env file or environment variables.
# We keep token loading inside the backend helper so the code is reusable later.
```

```python
# Import our small Mapbox downloader helper from the backend code.
# Keeping download logic in backend/app makes it reusable later in FastAPI.
from backend.app.imagery.mapbox_static import download_mapbox_static_image
```

```python
# Set the map location and image options for this test.
# Selected Vienna coordinates: latitude 48.18461202879178, longitude 16.400399172025814.
lat = 48.18461202879178
lon = 16.400399172025814

# Zoom 16 covers a wider area than zoom 18, which is useful for route-level context.
# Notebook 03 assumes this same zoom when converting image pixels to lon/lat.
zoom = 16

# Mapbox Static Images uses a base size plus @2x retina scaling in our helper.
# width=512 and height=512 produce a final 1024x1024 pixel image.
image_width = 512
image_height = 512

# This is where the downloaded image will be saved.
image_path = project_root / "data" / "mapbox_satellite_vienna.png"
```

```python
# Download the satellite image from Mapbox and save it to disk.
# The helper creates the data folder automatically if needed.
saved_path = download_mapbox_static_image(
    lat=lat,
    lon=lon,
    zoom=zoom,
    width=image_width,
    height=image_height,
    output_path=str(image_path),
)

print(f"Saved satellite image to: {saved_path}")
```

```python
# Display the saved image in the notebook.
# matplotlib is a common Python library for showing images and charts.
import matplotlib.pyplot as plt
from PIL import Image

# Open the image file that we saved in the previous cell.
image = Image.open(saved_path)

# Confirm the final image size. It should be 1024x1024 because we requested 512x512@2x.
print(f"Downloaded image size: {image.size[0]} x {image.size[1]} pixels")

# Make the displayed image large enough to inspect.
plt.figure(figsize=(8, 8))
plt.imshow(image)

# Hide the x/y axes because this is an image, not a chart.
plt.axis("off")

# Add a simple title so we know what we are looking at.
plt.title("Mapbox Satellite Image - Vienna Center")
plt.show()
```
